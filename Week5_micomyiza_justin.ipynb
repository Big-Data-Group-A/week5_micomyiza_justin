```python
temps =[24.4,25.9,28.6,19.3,26.0,22.1]
raw = ["24.4","25.9","28.6","19.3"]

convert_t = [t*9/5 +32 for t in temps]
print(convert_t)

# exrices 1.2 a and b
 #hot

hot =[ t for t in temps if t>=26]
print(f"hot days: ",hot)

#cool_rounded

cool_rounded = [round(t) for t in temps if t <= 24]
print(cool_rounded)

# convert types

raw_f = [float(x) for x in raw]
avg_r = [sum(raw_f) / len(raw_f)]
print(f"Parsed floats: {raw_f}")
print("Average :", f"{avg_r}")
```
