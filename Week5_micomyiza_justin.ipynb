```python
# numpy arrays
# Exercies 3.1 the speed test
import pandas as pd
import numpy as np
import time

df = pd.read_excel("week5_kigali_weather.xlsx")


dates = df["date"].to_numpy()
temps = df["temp_c"].to_numpy()
rain = df["rainfall_mm"].to_numpy()
hum = df["humidity_pct"].to_numpy()





numbers = list(range(1000000))
arr = np.array(numbers)
t0=time.time()
doubled=arr*2
print(f"numpy fast: {time.time()-t0:.3f} s")

# exercises 3.2 describe the temperature 
temps = np.array(temps)
print(f"{temps.mean():.2f} °C")
print(f"{temps.max():.2f} °C")
print(f"{temps.min():.2f} °C")
print(f"{temps.std():.2f} °C")


# Exercies 3.3 the hottest day
max_idx = np.argmax(temps)
max_idx = np.argmax(temps)
print(f"Hottest day was {dates[max_idx]} with a temperature of {temps[max_idx]} °C")
```

```python
# Exercise 4.1 — Hot days (3 marks)
hot_mask = temps >= 26
hot_count = hot_mask.sum()
hot_avg_temp = temps[hot_mask].mean()
print(f"Number of hot days (>=26°C): {hot_count}")
print(f"Average temperature on hot days: {hot_avg_temp:.2f} °C")

# Exercise 4.2 — Rain report (3 marks)
rainy_mask = rain > 0
rainy_days_count = rainy_mask.sum()
total_rainfall = rain.sum()
max_rainfall = rain.max()
print(f"Number of rainy days: {rainy_days_count}")
print(f"Total rainfall over 60 days: {total_rainfall:.2f} mm")
print(f"Heaviest single-day rainfall: {max_rainfall:.2f} mm")

# Exercise 4.3 — Cross-array analysis (4 marks)
dry_mask = rain == 0
avg_temp_rainy = temps[rainy_mask].mean()
avg_temp_dry = temps[dry_mask].mean()
hot_and_rainy_count = ((temps >= 26) & (rain > 0)).sum()
high_humidity_count = (hum >= 70).sum()

print(f"Average temp on rainy days: {avg_temp_rainy:.2f} °C")
print(f"Average temp on dry days: {avg_temp_dry:.2f} °C")
print(f"Number of hot AND rainy days: {hot_and_rainy_count}")
print(f"Number of days with humidity >= 70%: {high_humidity_count}")
```

```python
# Exercise 5.1 — Anomaly detection (4 marks)
anomaly = temps - temps.mean()

hot_anomaly_mask = anomaly > 3
cold_anomaly_mask = anomaly < -3

print("Dates of days more than 3°C above mean:", dates[hot_anomaly_mask])
print("Count of days more than 3°C below mean:", cold_anomaly_mask.sum())

# Exercise 5.2 — Is the season changing? (3 marks)
first_30_avg = temps[:30].mean()
last_30_avg = temps[30:].mean()

print(f"First 30 days average temp: {first_30_avg:.2f} °C")
print(f"Last 30 days average temp: {last_30_avg:.2f} °C")

# Comment on trend:
# If last_30_avg > first_30_avg:
# # Kigali got warmer over the course of the 60 days.
# else:
# # Kigali got cooler over the course of the 60 days.
```

```python
paired_rain = list(zip(dates, rain))
top_5_wettest = sorted(paired_rain, key=lambda x: x[1], reverse=True)[:5]

print("Top 5 wettest days:")
for date, amount in top_5_wettest:
    print(f"Date: {date} | Rainfall: {amount:.2f} mm")
```
